### Define Scam Category ###

```python
scam_keywords = {
    'lottery': ['lottery', 'won', 'winner', 'prize', 'congratulations', 'claim your', 'jackpot'],
    'banking_kyc': ['kyc', 'account will be blocked', 'account suspended', 'update your account', 'bank account', 'debit card', 'credit card blocked', 'verify your account'],
    'phishing': ['click here', 'verify your account', 'login to your', 'confirm your identity', 'suspicious activity', 'security alert'],
    'job_scam': ['work from home', 'earn $', 'no experience required', 'hiring now', 'job offer', 'per week', 'easy money'],
    'investment': ['investment opportunity', 'guaranteed returns', 'double your money', 'crypto', 'bitcoin', 'trading profit', 'forex'],
    'upi_payment': ['upi', 'google pay', 'phonepe', 'paytm', 'refund', 'payment failed', 'transaction failed'],
    'tech_support': ['virus detected', 'your computer', 'microsoft support', 'tech support', 'call this number immediately', 'system infected'],
    'digital_arrest': ['digital arrest', 'police', 'cbi', 'court notice', 'legal action', 'arrest warrant', 'customs department', 'income tax department']
}
```

```python
def assign_scam_type(text):
    text = str(text).lower()
    for category, keywords in scam_keywords.items():
        for kw in keywords:
            if kw in text:
                return category
    return 'unknown'
```

```python
df_sms['scam_type'] = df_sms.apply(
    lambda row: assign_scam_type(row['message']) if row['label'] == 'scam' else 'none',
    axis=1
)

df_sms['scam_type'].value_counts()
```

```python
df_phish = pd.read_csv(r'../data/processed/phishing_email_clean.csv')
df_phish['scam_type'] = df_phish['label'].apply(lambda x: 'phishing' if x == 'scam' else 'none')
df_phish['scam_type'].value_counts()
```

```python
df_jobs2 = pd.read_csv('../data/processed/job_postings_clean.csv')
df_jobs2['scam_type'] = df_jobs2['label'].apply(lambda x: 'job_scam' if x == 'scam' else 'none')
df_jobs2['scam_type'].value_counts()
```

```python
df_fraud = pd.read_csv('../data/processed/fraud_email_clean.csv')
df_fraud['scam_type'] = df_fraud.apply(
    lambda row: assign_scam_type(row['message']) if row['label'] == 'scam' else 'none',
    axis=1
)
df_fraud['scam_type'].value_counts()
```

```python
df_phish.to_csv('../data/processed/phishing_email_clean.csv', index=False)
df_jobs.to_csv('../data/processed/job_postings_clean.csv', index=False)
df_fraud.to_csv('../data/processed/fraud_email_clean.csv', index=False)
df_sms.to_csv('../data/processed/sms_clean.csv', index=False)
```
